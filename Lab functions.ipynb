```python
#Question no 1:
from datetime import datetime

def greet_with_time(name):
    current_time = datetime.now().strftime("%H:%M:%S")
    print(f"Hello, {name}! The current time is {current_time}.")
greet_with_time("Fatima")
```

```python
#Question no 2:
def calculate_average(num1, num2, num3):
    average = (num1 + num2 + num3) / 3
    return average
result = calculate_average(10, 20, 30)
print("Average:", result)
```

```python
#Question no 3:
def calculate_discount(price, discount_percentage=10):
    discount_amount = (price * discount_percentage) / 100
    final_price = price - discount_amount
    return final_price
print(calculate_discount(100))      
print(calculate_discount(100, 20))  
```

```python
#Question no 4
def print_info(name, age, city):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")
print_info(name="Fatima", age=21, city="Lahore")
```

```python
#Question no 5:
def find_max(*args):
    return max(args)
print(find_max(3, 7, 2, 9, 5))
```

```python
#Question no 6:
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
```

```python
#Question no 7:
def apply_function(func, numbers):
    result = []
    for num in numbers:
        result.append(func(num))
    return result
def square(x):
    return x * x
numbers = [1, 2, 3, 4, 5]
print(apply_function(square, numbers))
```

```python
#Question no 8:
def apply_operation(func, numbers):
    return [func(num) for num in numbers]
def double(x):
    return x * 2
numbers = [1, 2, 3, 4, 5]
result = apply_operation(double, numbers)

print(result)
```

```python
#Question no 9:
import time
def measure_time(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        execution_time = end_time - start_time
        print(f"Execution time: {execution_time:.6f} seconds")
        return result
    return wrapper
@measure_time
def sample_function():
    time.sleep(2)
    print("Function executed")

sample_function()
```
