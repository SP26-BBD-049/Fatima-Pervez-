```python
#Question no 07:
# Take input from user
num = int(input("Enter a number: "))
num = abs(num)
count = 0
if num == 0:
    count = 1
else:
    while num > 0:
        num //= 10  
        count += 1
print("Number of digits:", count)
```

```python
#Question no 10:
secret = 25
attempts = 0
while True:
    guess = int(input("Guess the number: "))
    attempts += 1
    if guess > secret:
        print("Too high")
    elif guess < secret:
        print("Too low")
    else:
        print("Correct! You guessed it in", attempts, "attempt(s).")
        break
```
